## Research Questions and Answers

For every answer, distinguish between **known evidence**, **reasonable interpretation**, and **unknown information to investigate**.

### 1. Where did the data come from?

**Answer:** [**Known evidence** The dataset is "Student Stress Factors: A Comprehensive Analysis" (1,100 rows, 21 columns), which is part of the "Student Stress Monitoring Datasets" collection on Kaggle and published by Md Sultanul Islam Ovi. The data is self-reported, so it reflects how students perceive their own stress. **Reasonable interpretation:** Self reported ratings can be influenced by mood and by how each student understands the questions. **Unknown:** Who originally collected the data, at which institutions and in which country, when it was collected, and how.]

### 2. Why was it originally collected?

**Answer:** [**Known evidence** The dataset was made to study student stress and well-being. **Reasonable interpretation:** Our project uses it for the same broad purpose. The data is anonymous, so we judge the ethical concern from the change of purpose to be low. However, it is not zero because the data covers sensitive topics such as mental health and anxiety. **Unknown:** Whether students consented to their answers being publicly shared and reused, and whether the original survey had ethics approval.]

### 3. Who is represented?

**Answer:** [**Known evidence** Students, described as covering psychological, environmental, academic, and social stress factors. **Reasonable interpretation:** Because the data is about students, the results say something about student stress in general, but only for the kind of students who were surveyed. **Unknown:** The age range, gender split, universities, countries and time period.]

### 4. Who might be missing?

**Answer:** [**Known evidence** There is no documentation that describes the characteristic of students in this sample, other than the fact that they are students. **Reasonable interpretation:** A survey of this kind is likely to miss students who did not respond, such students may have little time or lack internet access. **Unknown:** The response rate to this survey and who declined to take part.]

### 5. What personal or sensitive information does it contain, explicitly or implicitly?

**Answer:** [**Known evidence:** The dataset has 21 columns, including self reported ratings of anxiety_level, self_esteem, mental_health_history, depression, headache, blood_pressure, sleep_quality, breathing_problem, noise_level, living_conditions, safety, basic_needs, academic_performance, study_load, teacher_student_relationship, future_career_concerns, social_support, peer_pressure, extracurricular_activities, bullying, and stress_level. There are no direct identifiers such as names or emails. **Reasonable interpretation:** These factors can be deemed as personal and sensitive information. **Unknown:** Whether the students were from a single institution.]

### 6. What could your analysis reveal that was not obvious in the original data?

**Answer:** [**Known evidence:** Related studies using this data report links between stress and factors such as anxiety, depression, self-esteem, sleep quality, and bullying. **Reasonable interpretation:** Our analysis could show which factors are most strongly related to stress, and which factors are related to each other.  **Unknown:** How stress_level was defined and measured, whether it is reliable, and whether the patterns we find would hold for other groups of students.]

### 7. Who could benefit from your findings?

**Answer:** [**Known evidence:** The dataset was published to support the study of student stress and its causes. **Reasonable interpretation:** Students, professors, academic staff, parents of students, and friends of students. The findings may help students in the future to handle college-related stress more efficiently and perform better academically. They may also give academic staff useful insight into how present-day students are coping with college, and could help university counselling services decide where to focus support. **Unknown:** Whether the results would generalise beyond this sample, and whether any institution would actually use them.]

### 8. Who could be harmed or misrepresented?

**Answer:** [**Known evidence:** The data covers sensitive topics, including mental health history, depression, anxiety, and bullying. **Reasonable interpretation:** Students could be harmed if results were used to label individuals as "high stress" or "at risk". Students with mental health difficulties, students experiencing bullying, and those with poor living conditions are the most vulnerable if the data were ever linked back to them. Students who are not represented in the sample, such as mature or postgraduate students, could be misrepresented if our conclusions were applied to all students. **Unknown:** Whether a model built on this data would work equally well for different groups of students, and who the sample students actually are.]
